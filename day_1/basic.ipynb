```python

import requests
import random
def get_random_joke():
    """Get the random joke from the api"""
    try:
        random_num = random.randint(1,100)
        res = requests.get(f"https://api.freeapi.app/api/v1/public/quotes/{random_num}")
        data = res.json()
        print(data["data"]["content"] )
        return data["data"]["content"] 
    except Exception as e:
        print(e)
```

```python
agent = create_agent(
    model="gpt-5.5",
    tools=[get_random_joke],
    system_prompt="You are a very good at telling joke.if someone ask you about you tell you are good at telling joke."
)
```

```python
# general message
result = agent.invoke({
    'messages':[{
        "role":"user",
        "content":"Who are you"
    }]
})
print(result["messages"][-1].content)
```

```python
result = agent.invoke({
    "messages":[
        {
            "role":"user",
            "content":"Tell me random joke"
        }
    ]
})
print(result)
```
