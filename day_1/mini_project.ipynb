```python
"""
Generate the content by llm
store the content into any types of file with desire file name
"""
```

```python
from langchain.agents import create_agent
import os 
from dotenv import load_dotenv 

load_dotenv()
```

```python
# just create the function to convert the content into file
def convert_into_file(content,filetype,filename):
    """ 
    Save the file into any formate as per requires.
    Args:
    content: The text which need to written in file
    filetype: Extention of file(eg.txt,pdf)
    filename: Name of file in which we need to save the file
    """
    filename = filename + "." + filetype
    with open(filename,'w') as file:
        file.write(content)
    return f"output is save with this name {filename}"

```

```python
# creating the agent

system_prompt = """
You are smart assistent.
If the user as for saving the content into file then call the tools and save it.
 """
agent = create_agent(
    model="gpt-5.5",
    system_prompt=system_prompt,
    tools=[convert_into_file]
)
```

```python
result = agent.invoke({
    "messages":[{"role":"user",
    "content":"Who are you ?"}]
})

print(result["messages"][-1].content)
```

```python
prompt = "Write someting about football and save the file into pdf formate with file name of football"
result = agent.invoke({
    "messages":[{
        "role":"user",
        "content":prompt
    }]
})
print(result)
```
