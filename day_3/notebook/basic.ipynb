```python
from pathlib import Path
```

```python
x = Path("../data/pdf")
lst = list(x.glob('*.pdf'))
```

```python
def process_all_pdf(path):
    all_doc = []
    path = Path(path)
    filepath_lst = list(path.glob("*.pdf"))
    print(f"Total number of file to process are {len(filepath_lst)} \n")

    for filepath in filepath_lst:
        print(f"Processing:{filepath.name}")
        loader = PyMuPDFLoader(file_path=filepath)
        data = loader.load()
        print("Total pages:",len(data), "\n")
        all_doc.extend(data)
    return all_doc
```

```python
document = process_all_pdf("../data/pdf")
print(document)
```

### Chunking 

```python
def chunk_document(document,chunk_size=500,chunk_overlap=100):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        separators=["\n\n","\n",""," "]
        )
    chunk = text_splitter.split_documents(documents=document)
    return chunk
```

```python
chunk = chunk_document(document)
```
